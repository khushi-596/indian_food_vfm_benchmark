```python
import json, random
from collections import Counter, defaultdict

SPL = f"{PROJECT_DIR}/splits"
exclude  = set(json.load(open(f"{SPL}/dedup_exclude.json")))
group_of = json.load(open(f"{SPL}/group_ids.json"))
targets  = list(dataset.targets)
assert len(targets) == len(group_of) == 4578, 'dataset / dedup files mismatch. STOP.'

used = [i for i in range(len(targets)) if i not in exclude]
n_c = Counter(targets[i] for i in used)
assert min(n_c.values()) >= 3
quota = {c: {'test': max(1, round(0.15 * n)), 'val': max(1, round(0.15 * n))} for c, n in n_c.items()}
for c in quota:
    quota[c]['train'] = n_c[c] - quota[c]['test'] - quota[c]['val']

groups = defaultdict(list)
for i in used:
    groups[group_of[i]].append(i)
rng = random.Random(42)
order = sorted(groups.values(), key=lambda g: (-len(g), rng.random()))   # big groups first, then random

cur = {c: Counter() for c in n_c}
assign = {'train': [], 'val': [], 'test': []}
for g in order:
    cls = Counter(targets[i] for i in g)
    best, best_score = None, None
    for s in ('test', 'val', 'train'):
        if all(cur[c][s] + k <= quota[c][s] for c, k in cls.items()):
            score = sum((quota[c][s] - cur[c][s]) / quota[c][s] for c in cls if quota[c][s] > 0)
            if best_score is None or score > best_score:
                best, best_score = s, score
    if best is None:   # no split has room for the whole group: the one with most room left
        best = max(('train', 'val', 'test'), key=lambda s: sum(quota[c][s] - cur[c][s] for c in cls))
    assign[best] += g
    for c, k in cls.items():
        cur[c][best] += k

train_idx, val_idx, test_idx = assign['train'], assign['val'], assign['test']
for lst in (train_idx, val_idx, test_idx):
    rng.shuffle(lst)

assert not (set(train_idx) & set(val_idx)) and not (set(train_idx) & set(test_idx)) and not (set(val_idx) & set(test_idx))
for name, lst in [('train', train_idx), ('val', val_idx), ('test', test_idx)]:
    cnt = Counter(targets[i] for i in lst)
    assert len(cnt) == 100, f'some class has no {name} image'
    print(f'{name:5s}: {len(lst)} images | per class min {min(cnt.values())} max {max(cnt.values())}')
gs = defaultdict(set)
for name, lst in [('train', train_idx), ('val', val_idx), ('test', test_idx)]:
    for i in lst:
        gs[group_of[i]].add(name)
assert all(len(v) == 1 for v in gs.values()), 'a near-duplicate group was split'
print(f'{len(exclude)} images excluded as duplicates; {len(used)} used; all near-duplicate groups kept within one split')
```

```python
from torchvision.models import (
    resnet50,
    ResNet50_Weights
)

num_classes = len(dataset.classes)

model = resnet50(
    weights=ResNet50_Weights.IMAGENET1K_V2
)

model.fc = nn.Linear(
    model.fc.in_features,
    num_classes
)

model = model.to(device)

print(model)

print(
    "\nNumber of Classes:",
    num_classes
)
```

```python
total_params = sum(
    p.numel()
    for p in model.parameters()
)

trainable_params = sum(
    p.numel()
    for p in model.parameters()
    if p.requires_grad
)

print(
    f"Total Parameters: {total_params:,}"
)

print(
    f"Trainable Parameters: {trainable_params:,}"
)
```

```python
from sklearn.utils.class_weight import compute_class_weight

weights = compute_class_weight(

    class_weight="balanced",
    classes=np.unique(dataset.targets),
    y=dataset.targets
)

weights = torch.tensor(
    weights,
    dtype=torch.float
).to(device)

criterion = nn.CrossEntropyLoss(
    weight=weights
)
```

```python
optimizer = optim.AdamW(
    model.parameters(),
    lr=3e-4,
    weight_decay=1e-4
)
```

```python
scheduler = optim.lr_scheduler.CosineAnnealingLR(
    optimizer,
    T_max=20
)
```

```python
scaler = torch.cuda.amp.GradScaler()
```

```python
best_val_acc = 0
patience = 7
counter = 0
```

```python
EPOCHS = 35
train_losses = []
val_losses = []
train_accs = []
val_accs = []
```

```python
def train_epoch():

    model.train()
    running_loss = 0
    correct = 0
    total = 0

    for images, labels in train_loader:
        images = images.to(device)
        labels = labels.to(device)
        optimizer.zero_grad()
        with torch.cuda.amp.autocast():
            outputs = model(images)
            loss = criterion(
                outputs,
                labels
            )

        scaler.scale(loss).backward()
        scaler.step(optimizer)
        scaler.update()
        running_loss += loss.item()
        preds = outputs.argmax(1)
        correct += (
            preds == labels
        ).sum().item()

        total += labels.size(0)

    acc = correct / total

    loss = (
        running_loss /
        len(train_loader)
    )

    return loss, acc
```

```python
def validate():

    model.eval()
    running_loss = 0
    correct = 0
    top5_correct = 0
    total = 0

    with torch.no_grad():

        for images, labels in val_loader:
            images = images.to(device)
            labels = labels.to(device)
            outputs = model(images)
            loss = criterion(
                outputs,
                labels
            )

            running_loss += loss.item()
            preds = outputs.argmax(1)

            correct += (
                preds == labels
            ).sum().item()

            _, top5 = outputs.topk(
                5,
                dim=1
            )

            top5_correct += (
                top5 ==
                labels.view(-1,1)
            ).sum().item()

            total += labels.size(0)

    loss = running_loss / len(val_loader)
    acc = correct / total
    top5_acc = top5_correct / total
    return loss, acc, top5_acc
```

```python
for epoch in range(EPOCHS):

    train_loss, train_acc = train_epoch()
    val_loss, val_acc, val_top5 = validate()
    scheduler.step(val_loss)

    train_losses.append(
        train_loss
    )

    val_losses.append(
        val_loss
    )

    train_accs.append(
        train_acc
    )

    val_accs.append(
        val_acc
    )

    print(
        f"Epoch {epoch+1}/{EPOCHS}"
    )

    print(
        f"Train Loss: {train_loss:.4f}"
    )

    print(
        f"Val Loss: {val_loss:.4f}"
    )

    print(
        f"Train Acc: {train_acc:.4f}"
    )

    print(
        f"Val Acc: {val_acc:.4f}"
    )

    if val_acc > best_val_acc:

        best_val_acc = val_acc

        counter = 0

        torch.save(
            model.state_dict(),
            f"{PROJECT_DIR}/models/best_resnet50.pth"
        )

        print(
            f"Best Model Saved (Val Acc={val_acc:.4f})"
        )

    else:

        counter += 1

    if counter >= patience:

        print(
            "Early Stopping"
        )

        break
    print(
        f"Val Top5: {val_top5:.4f}"
    )
```
